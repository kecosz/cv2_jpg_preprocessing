```python
import numpy as np
import matplotlib.pyplot as plt
#from keras.preprocessing import image
#from keras.preprocessing.image import ImageDataGenerator,array_to_img,img_to_array
import cv2
from google.colab.patches import cv2_imshow

import glob
files = glob.glob("/content/drive/MyDrive/data/jpg111523/*") #google driveの"/data/jpg111523"フォルダ格納したファイルのリスト
files
```

```python
im_gauss_s = []
for n,i in enumerate(files):
  im = cv2.imread(i)
  im_gauss = cv2.GaussianBlur(im, (5, 5), 50) #!パラメータ
  imgs = cv2.hconcat([im, im_gauss])
  cv2_imshow(imgs)
  im_gauss_s.append(im_gauss)
```
